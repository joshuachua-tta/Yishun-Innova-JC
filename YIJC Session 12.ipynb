```python
# open text file, choose language crtl + k to be python, open terminal, click arrow down on the top, choose command prompt, install pip opencv-python
# structure of a python file, this is not a notebook, code runs top down
# structure is import first, as well as defining the variables
# write ur main loop, pip 8 guideline

#Imports
import cv2 #computer vision, c is a different language
import numpy as np #number python
from ugot import ugot 

got = ugot.UGOT() #why? creating one instance of my ugot
got.initialize("192.168.88.1")
got.open_camera()
print("Success!")

#Speed ranges
MOVE_MIN, MOVE_MAX = 5, 80 #anything not changeable is in caps, while the lowercase is changeable
TURN_MIN, TURN_MAX = 5, 280 #this value is not arbitrary

def clamp(val, lo, hi): #clamp prevents sprite from moving away from the screen
    return max(lo, min(hi, val))

#arm values


# Main Loop
def main(): #considered good practise to name ur loop main
    move_speed = 30 #initial move speed, you can change this later
    turn_speed = 45 # initial turn speed
    arm_x = 0 #left/right movement of j1
    arm_y = 0 #up/down movement of j2/j3


    while True: #do this thing forever
        frame = got.read_camera_data()
        if not frame:
            print("Failed to grab frame") #fail safe
            break #only way to break out of forever loop

        #indentation is defined as scope, thing of surds indentations
        nparr = np.frombuffer(frame, np.uint8) #it is not sent as a literal image, an array is a list [n1, n2,...,nn]
        data = cv2.imdecode(nparr, cv2.IMREAD_COLOR) #off loading the work to the laptop, sent over data stream of nparr to laptop

        # Overlay current speeds
        info = f"Move speed: {move_speed}    | Turn speed: {turn_speed}"
        cv2.putText(
            data, #downloading shaders ahh
            info, 
            (10, 30), #top left of your screen is the origin point
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8, #80% size
            (0, 255, 0), #green
            2,
            cv2.LINE_AA,
        )

        cv2.imshow("Robot Feed", data) #im is image, robot feed is actually the image display

        #Press 'q' to quit
        key = cv2.waitKey(1) & 0xFF

        # --- movement with WASD using speeds ---
        if key == ord("w"):
            got.mecanum_move_speed(0, move_speed) #forward
        elif key == ord("s"):
            got.mecanum_move_speed(1, move_speed) #backward
        elif key == ord("a"):
            got.mecanum_turn_speed(2, turn_speed) #left
        elif key == ord("d"):
            got.mecanum_turn_speed(3, turn_speed) #right
        elif key == ord(" "): #space to stop
            got.mecanum_stop()
        elif key == ord("q"): # q to quit
            break

        # --- arm movement with IJKL using variables ---
        #LEFT/RIGHT
        if key == ord("j"):
            arm_x += 5
            got.mechanical_joint_control(arm_x, arm_y, arm_y, 100) #joint 1/2/3/ duration
        elif key == ord("l"):
            arm_x -= 5
            got.mechanical_joint_control(arm_x, arm_y, arm_y, 100)

        #UP/DOWN
        elif key == ord("i"):
            arm_y += 5
            got.mechanical_joint_control(arm_x, arm_y, arm_y, 100)
        elif key == ord("k"):
            arm_y -= 5
            got.mechanical_joint_control(arm_x, arm_y, arm_y, 100)          



        #if cv2.waitKey(1) & 0xFF == ord('q'): #checks if you click the key in miliseconds, in this case it is 1
            #print("Done")
            #break

    cv2.destroyAllWindows() #memory leak if u dont do this, ran externally from your file

if __name__ == "__main__": #surrounded underscores is called dunder, dunder methods are specialised, specifically checking the hardcoded name of the file you're typing from
    main() 





        

```
