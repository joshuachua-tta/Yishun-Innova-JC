# Adding a Timer (+ Score) — Teaching Reference

## Why the timer needs to start on Play, not at launch
If you start counting time at the top of the script, time spent sitting on the
menu screen gets counted too. Start the clock the moment `game_state` switches
to `"playing"`.

```python
if event.type == pygame.MOUSEBUTTONDOWN and game_state == "menu":
    if play_button.collidepoint(event.pos):
        game_state = "playing"
        start_ticks = pygame.time.get_ticks()  # timer starts now
```

---

## Step 1 — Calculate elapsed time each frame
Put this near the top of the `"playing"` block, alongside the other update logic.

```python
elapsed_seconds = (pygame.time.get_ticks() - start_ticks) // 1000
```

- `pygame.time.get_ticks()` returns milliseconds since `pygame.init()` was called.
- Subtracting `start_ticks` gives milliseconds *since Play was pressed*.
- `// 1000` converts milliseconds to whole seconds.

---

## Step 2 — Render and draw the timer text
Reuse the `font` object already created near the top of the file
(`font = pygame.font.SysFont(None, 36)`).

Put this in the drawing section — after `screen.fill(...)`, before
`pygame.display.flip()`.

```python
timer_text = font.render(f"Time: {elapsed_seconds}s", True, WHITE)
screen.blit(timer_text, (10, 10))
```

---

## Step 3 — (Bonus) Add the score display the same way
Since `SCORE` is already being tracked on collision, showing it is just one more
`render` + `blit` pair:

```python
score_text = font.render(f"Score: {SCORE}", True, WHITE)
screen.blit(score_text, (10, 50))
```

## Common mistakes to watch for
- **Timer keeps running from menu screen** → `start_ticks` was set at script
  start instead of when Play was clicked.
- **Text renders but doesn't show** → `blit()` called *before* `screen.fill()`
  gets drawn over immediately after.
- **`NameError: start_ticks not defined`** → happens if a student jumps
  straight into `"playing"` state code without going through the Play button
  click first, or if `start_ticks` was only set inside the `if` block and
  never initialized elsewhere.
- **Timer shows huge/wrong number on first frame of "playing"** → make sure
  `start_ticks` is being reset every time Play is clicked, not just once.
