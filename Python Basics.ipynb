### Rules for Naming Variables

2. Cannot start with a number.

1. Avoid capitalizing names unless that naming convention is constant throughout the program
2. Name the variable in a way that represents the data that will be assigned (e.g. name = **"Sebastian"**, school = **"YIJC"**)
3. Name the variable in a simple yet consise manner (e.g. **student_name** can be **std_name**, but not too simple like **stdnm**)

## 1. Talking to the User: `print()` and `input()`
The `print()` function sends data to the screen, while `input()` allows the user to type data into the program.

**Crucial Note:** The `input()` function always returns a **String** (text), even if the user types a number.

## 2. The Transformers: Type Casting

## 2. Advanced String Manipulation
Strings in Python are more than just text; they are sequences of characters. This means we can perform "math" on them and cut them into pieces.

### Concatenation and Repetition
- **Concatenation (`+`):** Glues two strings together.
- **Repetition (`*`):** Repeats a string multiple times.

```python
laugh = "Ha"
print(laugh * 3) # Result: HaHaHa

greeting = "Hello" + " " + "Class"
print(greeting)
```

### Slicing: Cutting Strings
You can extract specific parts of a string using brackets `[]`. The syntax is `[start:stop]`. 
Remember: Python starts counting at **0**, and the `stop` index is **not included**.

```python
word = "Pythonism"

# Get the first 6 letters
print(word[0:6]) 

# Get letters from index 6 to the end
print(word[6:])
```

Note: The letter 'n' is the 6th letter in the string, which is index **5**

---
This chapter covers the "logic" of programming. We will learn how to make decisions using `if` statements and how to repeat tasks efficiently using `for` and `while` loops.

## 1. Conditional Statements (Decision Making)
Programs often need to make choices. We use `if`, `elif` (else if), and `else` to control which code runs based on a condition.

### Comparison Operators
Before making a decision, we need to compare values:

| Operator | Meaning | Example |
| :--- | :--- | :--- |
| `==` | Equal to | `5 == 5` (True) |
| `!=` | Not equal to | `5 != 3` (True) |
| `>` | Greater than | `10 > 2` (True) |
| `<` | Less than | `2 < 10` (True) |
| `>=` | Greater than or equal to | `5 >= 5` (True) |
| `<=` | Less than or equal to | `3 <= 5` (True) |

```python
age = 18

if age >= 18:
    print("You are an adult.")
elif age >= 13:
    print("You are a teenager.")
else:
    print("You are a child.")
```

## 2. Logical Operators
We can combine multiple conditions using `and`, `or`, and `not`.

- `and`: Both conditions must be True.
- `or`: At least one condition must be True.
- `not`: Reverses the result (True becomes False).

```python
has_ticket = True
is_invited = False

if has_ticket or is_invited:
    print("Welcome to the event!")
else:
    print("Access denied.")
```

Note: **`if has_ticket`** and **`if (has_ticket == True)`** both achieve the same result as True is assigned to the variable

## 3. For Loops (Repeating for a sequence)
Use a `for` loop when you know how many times you want to repeat a block of code, or when you want to go through a list/range.

```python
print("Counting 0 to 4:")
for i in range(5):
    print(i)
```

```python
print("\nIterating through a list:")
students = ["Alice", "Bob", "Charlie"]
for name in students:
    print(f"Hello, {name}!")
```

### The range() Function

The `range()` function is often the "engine" of a `for` loop. While we often use it with just one number (the `stop` value), it actually supports up to three arguments that give you precise control over your iterations.

### The Syntax
The full structure is `range(start, stop, step)`.

`start`: The starting value of the sequence (inclusive). If omitted, it defaults to **0**.

`stop`: The number where the sequence ends (**exclusive**). The loop will stop before it reaches this number.

`step`: The "jump" between each number. If omitted, it defaults to **1**.

| Goal | Code | Resulting Sequence |
| :--- | :--- | :--- |
| **Standard** | `range(5)` | `0, 1, 2, 3, 4` |
| **Custom Start** | `range(2, 6)` | `2, 3, 4, 5` |
| **With Step** | `range(0, 10, 2)` | `0, 2, 4, 6, 8` |
| **Counting down** | `range(5, 0, -1)` | `5, 4, 3, 2, 1` |

# Chapter 5: The Art of Debugging 🔍
---
Programming isn't about writing perfect code on the first try; it's about being a good detective when things go wrong. This chapter will teach you how to identify, understand, and fix errors in your Python programs.

## 1. Types of Errors
Errors (often called **bugs**) generally fall into three categories:

| Error Type | Description | When does it happen? |
| :--- | :--- | :--- |
| **Syntax Error** | A "grammar" mistake in the code (missing colon, unclosed quotes). | Before the code even runs. |
| **Runtime Error** | The code starts running but crashes halfway (dividing by zero, calling a variable that doesn't exist). | While the code is executing. |
| **Logic Error** | The code runs perfectly and doesn't crash, but the result is wrong. | While the code is executing. |

## 2. Reading a Traceback
When Python crashes, it provides a **Traceback**. Most people panic and ignore it, but the information you need is usually at the very bottom.

- **NameError**: You used a variable name that hasn't been defined yet.
- **TypeError**: You tried to perform an operation on the wrong type (e.g., adding a string to an integer).
- **IndexError**: You tried to access an item in a list that doesn't exist (e.g., item 10 in a list of 3).

```python
# Try to run this cell and look at the error message at the bottom
print(10 / 0)
```
