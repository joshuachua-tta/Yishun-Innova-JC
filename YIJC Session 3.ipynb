```python
import time
from ugot import ugot

got = ugot.UGOT()

got.initialize('192.168.1.88') #UGOT IP Address

print("Connection Successful")
```

```python
#Gesture test

got.load_models(["gesture"])
while True:
    gesture = got.get_gesture_result()
    print(f"{gesture}          ", end='\r')
    time.sleep(1)
```

## Rock Paper Scissors

```python
import random
got.play_audio_tts("Rock, Paper, Scissors, shoot",0,True)
gesture_list = ["rock", "paper", "scissors"]
robot = random.choice(gesture_list)
got.play_audio_tts(robot,0,True)
got.load_models(["gesture"])
gesture = ""
while True:
    gesture = got.get_gesture_result()
    if gesture == "Stone" or gesture == "Scissors" or gesture == "Cloth":
        break

if robot == "rock" and gesture == "Stone":
    got.play_audio_tts("We have a tie",0,True)
elif robot == "paper" and gesture == "Cloth":
    got.play_audio_tts("We have a tie",0,True)
elif robot == "scissors" and gesture == "Scissors":
    got.play_audio_tts("We have a tie",0,True)

elif robot == "rock" and gesture == "Scissors":
    got.play_audio_tts("I win!",0,True)
elif robot == "scissors" and gesture == "Cloth":
    got.play_audio_tts("I win!",0,True)
elif robot == "paper" and gesture == "Stone":
    got.play_audio_tts("I win!",0,True)

else:
    got.play_audio_tts("You win!",0,True)
```
